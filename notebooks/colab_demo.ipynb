最初のセルは、既定ではGitHubからコードを取得します。Googleドライブ上のフォルダで作業したい場合は `USE_DRIVE = True` にします。

```
import os, subprocess

USE_DRIVE = False   # True: Google Drive の作業フォルダを使う / False: GitHub からクローンする
DRIVE_DIR = "/content/drive/MyDrive/PAI応用/最終課題/pai-final-language-pick-place"
REPO_URL = "https://github.com/ryuseeei07-svg/pai-final-language-pick-place.git"
CLONE_DIR = "/content/pai-final-language-pick-place"

if USE_DRIVE:
    from google.colab import drive
    drive.mount("/content/drive")
    os.chdir(DRIVE_DIR)
else:
    if not os.path.isdir(CLONE_DIR):
        subprocess.run(["git", "clone", REPO_URL, CLONE_DIR], check=True)
    os.chdir(CLONE_DIR)
print(os.getcwd())
```
